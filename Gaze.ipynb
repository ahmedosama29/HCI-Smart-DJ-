```python
predictor = dlib.shape_predictor(r"C:\Users\saif\Desktop\Gaze Tracking\shape_predictor_68_face_landmarks.dat")
detector = dlib.get_frontal_face_detector()
```

```python
cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()

    if not ret:
        break

    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    faces = detector(gray)
    #faces

    for face in faces:
        x1 = face.left()
        y1 = face.top()
        x2 = face.right()
        y2 = face.bottom()
        cv2.rectangle(frame, (x1, y1), (x2, y2), (0, 255, 0), 2)
        landmarks = predictor(gray, face)

        for n in range(0, 68):
            x, y = landmarks.part(n).x, landmarks.part(n).y
            cv2.circle(frame, (x, y), 1, (0, 255, 0), -1)

    cv2.imshow("Face Landmarks", frame)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np
import dlib

cap = cv2.VideoCapture(0)

lower_Black_laser = np.array([0, 0, 0])
upper_Black_laser = np.array([180, 255, 50])

detector = dlib.get_frontal_face_detector()
predictor = dlib.shape_predictor("shape_predictor_68_face_landmarks.dat")

def mid_point(p1, p2):
    return int((p1.x + p2.x) / 2), int((p1.y + p2.y) / 2)

while True:
    ret, frame = cap.read()
    if not ret:
        print("Failed to grab frame")
        break

    frame = cv2.resize(frame, (640, 480))
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    faces = detector(gray)

    for face in faces:
        landmarks = predictor(gray, face)

        left_eye_region = np.array([(landmarks.part(n).x, landmarks.part(n).y) for n in range(36, 42)], np.int32)
        h,w,_=frame.shape
        mask=np.ones((h,w),np.uint8)*255
        cv2.fillPoly(mask, [left_eye_region], 255)

        left_eye = cv2.bitwise_and(gray, gray, mask=mask)
        min_x = np.min(left_eye_region[:, 0])
        max_x = np.max(left_eye_region[:, 0])
        min_y = np.min(left_eye_region[:, 1])
        max_y = np.max(left_eye_region[:, 1])

        gray_eye = left_eye[min_y:max_y, min_x:max_x]
        _, threshold_eye = cv2.threshold(gray_eye, 70, 255, cv2.THRESH_BINARY)
        threshold_eye = cv2.resize(threshold_eye,None,fx=5,fy=5)

        mask_hsv = cv2.inRange(cv2.cvtColor(cv2.merge([threshold_eye, threshold_eye, threshold_eye]), cv2.COLOR_BGR2HSV), lower_Black_laser, upper_Black_laser)
        mask_hsv = cv2.erode(mask_hsv, None, iterations=2)
        mask_hsv = cv2.dilate(mask_hsv, None, iterations=2)

        contours, _ = cv2.findContours(mask_hsv, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

        if contours:
            max_contour = max(contours, key=cv2.contourArea)
            x, y, w, h = cv2.boundingRect(max_contour)
            center_x = x + w // 2
            center_y = y + h // 2
            radius = max(w, h) // 2  

            cv2.circle(threshold_eye, (center_x, center_y), radius, (255, 180, 180), 2)


            cv2.circle(frame, (min_x + center_x, min_y + center_y), radius, (0, 255, 0), 2)

            coordinates_text = f"({min_x + x}, {min_y + y})"
            cv2.putText(frame, coordinates_text, (min_x + x, min_y + y - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (255, 255, 255), 1)


    cv2.imshow("Eye_Tracking", frame)
    cv2.imshow("threshold_eye",threshold_eye)


    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()

```

```python
import cv2
import numpy as np
import dlib

cap = cv2.VideoCapture(0)

lower_black = np.array([0, 0, 0])  
upper_black = np.array([180, 255, 50])  

detector = dlib.get_frontal_face_detector()
predictor = dlib.shape_predictor("shape_predictor_68_face_landmarks.dat")

while True:
    ret, frame = cap.read()
    if not ret:
        print("Failed to grab frame")
        break

    frame = cv2.resize(frame, (640, 480))
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    faces = detector(gray)

    for face in faces:
        landmarks = predictor(gray, face)

        left_eye_region = np.array([(landmarks.part(n).x, landmarks.part(n).y) for n in range(36, 42)], np.int32)
        h, w, _ = frame.shape
        mask = np.zeros((h, w), np.uint8)
        cv2.fillPoly(mask, [left_eye_region], 255)

        left_eye = cv2.bitwise_and(gray, gray, mask=mask)
        min_x = np.min(left_eye_region[:, 0])
        max_x = np.max(left_eye_region[:, 0])
        min_y = np.min(left_eye_region[:, 1])
        max_y = np.max(left_eye_region[:, 1])

        gray_eye = left_eye[min_y:max_y, min_x:max_x]
        _, threshold_eye = cv2.threshold(gray_eye, 70, 255, cv2.THRESH_BINARY_INV)
        threshold_eye = cv2.resize(threshold_eye,None,fx=5,fy=5)

        hsv_eye = cv2.cvtColor(cv2.merge([gray_eye, gray_eye, gray_eye]), cv2.COLOR_BGR2HSV)
        mask_hsv = cv2.inRange(hsv_eye, lower_black, upper_black)

        contours, _ = cv2.findContours(mask_hsv, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

        if contours:
            max_contour = max(contours, key=cv2.contourArea)
            (x, y), radius = cv2.minEnclosingCircle(max_contour)

            if radius > 5 and radius < 15:  
                center = (int(x), int(y))
                radius = int(radius)

                cv2.circle(threshold_eye, center, radius, (255, 255, 255), 2)
                cv2.circle(frame, (min_x + int(x), min_y + int(y)), radius, (0, 255, 0), 2)

                coordinates_text = f"({min_x + int(x)}, {min_y + int(y)})"
                cv2.putText(frame, coordinates_text, (min_x + int(x), min_y + int(y) - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (255, 255, 255), 1)

    cv2.imshow("Eye_Tracking", frame)
    cv2.imshow("threshold_eye", threshold_eye)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np
import dlib
import mediapipe as mp


cap = cv2.VideoCapture(0)

detector = dlib.get_frontal_face_detector()
predictor = dlib.shape_predictor("shape_predictor_68_face_landmarks.dat")

while True:
    ret, frame = cap.read()
    if not ret:
        print("Failed to grab frame")
        break

    frame = cv2.resize(frame, (640, 480))
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    faces = detector(gray)

    for face in faces:
        landmarks = predictor(gray, face)


        left_eye_region = np.array([(landmarks.part(n).x, landmarks.part(n).y) for n in range(36, 42)], np.int32)
        h, w = gray.shape
        mask = np.ones((h, w), np.uint8)*255
        cv2.fillPoly(mask, [left_eye_region], 255)

        left_eye = cv2.bitwise_and(gray, gray, mask=mask)
        min_x = np.min(left_eye_region[:, 0])
        max_x = np.max(left_eye_region[:, 0])
        min_y = np.min(left_eye_region[:, 1])
        max_y = np.max(left_eye_region[:, 1])

        gray_eye = left_eye[min_y:max_y, min_x:max_x]

        scaled_eye = cv2.resize(gray_eye, None, fx=5, fy=5, interpolation=cv2.INTER_CUBIC)

        _, binary_eye = cv2.threshold(gray_eye, 40, 255, cv2.THRESH_BINARY_INV)  
        contours, _ = cv2.findContours(binary_eye, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

        if contours:
            max_contour = max(contours, key=cv2.contourArea)
            (x, y), radius = cv2.minEnclosingCircle(max_contour)

            if radius > 2: 
                center = (int(x), int(y))
                radius = int(radius)

                cv2.circle(scaled_eye, (int(x * 5), int(y * 5)), radius * 5, (255, 255, 255), 2)  
                cv2.circle(frame, (min_x + center[0], min_y + center[1]), radius, (0, 255, 0), 2)  


                cv2.circle(scaled_eye, (int(x * 5), int(y * 5)), 3, (0, 0, 255), -1)  
                cv2.circle(frame, (int((x * 5)/(x * 5 *radius) * 640),int((y * 5)/y * 5 *radius) * 480), 3, (0, 0, 255), -1) 

                coordinates_text = f"({min_x + center[0]}, {min_y + center[1]})"
                cv2.putText(frame, coordinates_text, (min_x + center[0], min_y + center[1] - 10),
                            cv2.FONT_HERSHEY_SIMPLEX, 0.5, (255, 255, 255), 1)

    cv2.imshow("Eye_Tracking", frame)
    cv2.imshow("Grayscale Eye", scaled_eye)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
